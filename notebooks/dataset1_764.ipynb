```python
print("Supporting facts:")
for title, sent_id in zip(
    examples["supporting_facts"][0]["title"],
    examples["supporting_facts"][0]["sent_id"]
):
    print(f"- {title} → sentence {sent_id}")
```

```python
# maybe important
context = examples["context"][0]
supporting_facts = examples["supporting_facts"][0]

for title, sent_id in zip(
    supporting_facts["title"],
    supporting_facts["sent_id"]
):
    for passage_title, sentences in zip(context['title'], context['sentences']):
        if passage_title == title:
            print(f"\n===== SUPPORTING PASSAGE: {title} =====")
            print(f"[{sent_id}] {sentences[sent_id]}")
```
